# T01 — Baseline truyền thống cho Task A

Notebook này chỉ chạy script và hiển thị kết quả. Toàn bộ logic train, chọn cấu hình trên dev và đánh giá test nằm trong `ml/src/baselines.py`.

```python
from pathlib import Path
import subprocess
import sys

project_root = Path.cwd().resolve()
while project_root.parent != project_root and not (project_root / 'ml' / 'src' / 'baselines.py').is_file():
    project_root = project_root.parent
if not (project_root / 'ml' / 'src' / 'baselines.py').is_file():
    raise FileNotFoundError('Không tìm thấy thư mục gốc firefly-cinema')

command = [
    sys.executable, str(project_root / 'ml' / 'src' / 'baselines.py'),
    '--data', str(project_root / 'data' / 'uit-visfd' / 'processed'),
    '--out', str(project_root / 'ml' / 'results'),
]
subprocess.run(command, cwd=project_root, check=True)
```
